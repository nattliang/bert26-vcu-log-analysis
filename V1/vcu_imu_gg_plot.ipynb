# Calibrated IMU G-G Plot

This notebook builds a vehicle G-G diagram (lateral acceleration versus longitudinal acceleration) from a decoded `LOG####_imu.csv`. It:

1. finds a stationary calibration window,
2. rotates and scales the inferred gravity vector to **1 g downward**,
3. subtracts gravity and the stationary gyro bias,
4. estimates and displays the vehicle-forward direction, and
5. draws both a scatter G-G plot and a magma density heatmap, and
6. computes gap-aware, filtered, 250 ms sustained-g metrics and an improved log-density heatmap.

At rest, an accelerometer measures upward specific force, so the physical gravity vector is its negative. A single stationary pose can correct tilt, a common scale factor, and zero-rate gyro bias; it cannot identify independent per-axis accelerometer bias/scale or yaw. When speed data is available, yaw/forward is estimated by regressing horizontal IMU acceleration against `d(speed)/dt`. Set `FORWARD_IMU_VECTOR` when the sensor mounting direction is known.

## Configuration

The default `LOG0012` contains both stationary and driving data. `HEATMAP_SATURATION_COUNT = 8` means any bin containing eight or more samples uses the brightest end of the magma colormap. Keep this value between 5 and 10 for the requested low saturation point.


```python
DATA_DIR = Path("VCU/June11thTestingDataLog_AfterChargingBattery")
LOG_ID = "LOG0012"

# Use None for automatic selection, or specify elapsed seconds: (start_s, end_s).
CALIBRATION_TIME_RANGE_S = None
CALIBRATION_MAX_DURATION_S = 5.0
MIN_CALIBRATION_DURATION_S = 2.0
STATIONARY_SPEED_THRESHOLD_MPH = 0.5

# Optional known forward direction in raw IMU [x, y, z] coordinates.
# Example: np.array([1.0, 0.0, 0.0]) if raw IMU +x points forward.
FORWARD_IMU_VECTOR = None
MIN_SPEED_ACCEL_MPS2 = 0.15

HEATMAP_BINS = 75
HEATMAP_SATURATION_COUNT = 10  # requested range: 5-10 samples/bin
GG_LIMIT_G = None              # None selects a robust limit from the data
MAX_SCATTER_POINTS = 50_000

# Sustained-g analysis. These settings do not alter the original plotters below.
FILTER_CUTOFF_HZ = 5.0
SUSTAINED_WINDOW_S = 0.250
MIN_FILTER_SEGMENT_S = 1.0
MAX_SAMPLE_GAP_FACTOR = 3.0
ACCEL_SATURATION_G = 3.99
MOVING_SPEED_THRESHOLD_MPH = 1.0
ATTITUDE_STATIONARY_CORRECTION_TAU_S = 0.25
MAX_REASONABLE_GRAVITY_TILT_DEG = 20.0
IMPROVED_HEATMAP_BINS = 90
IMPROVED_GG_LIMIT_G = None  # None selects a robust limit from sustained data

ACCEL_COLS = ["accel_x_g", "accel_y_g", "accel_z_g"]
GYRO_COLS = ["gyro_x_dps", "gyro_y_dps", "gyro_z_dps"]
AXES = ["x", "y", "z"]
MPH_TO_MPS = 0.44704

if not 5 <= HEATMAP_SATURATION_COUNT <= 10:
    raise ValueError("HEATMAP_SATURATION_COUNT must be between 5 and 10")
```

```python
def find_repo_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for candidate in (start, *start.parents):
        if (candidate / "VCU").is_dir() and (candidate / "README.md").exists():
            return candidate
    raise FileNotFoundError("Could not find the repository root")


def resolve_data_dir(path):
    path = Path(path).expanduser()
    if path.is_absolute():
        return path
    repo_root = find_repo_root()
    for candidate in (Path.cwd() / path, repo_root / path, repo_root / "VCU" / path):
        if candidate.exists():
            return candidate.resolve()
    return (repo_root / path).resolve()


def load_imu_and_speed(data_dir, log_id):
    data_dir = resolve_data_dir(data_dir)
    imu_path = data_dir / f"{log_id}_imu.csv"
    if not imu_path.exists():
        raise FileNotFoundError(f"IMU log not found: {imu_path}")

    imu = pd.read_csv(imu_path)
    required = ["t_ms", *ACCEL_COLS, *GYRO_COLS]
    missing = [column for column in required if column not in imu.columns]
    if missing:
        raise ValueError(f"{imu_path.name} is missing columns: {missing}")

    for column in required:
        imu[column] = pd.to_numeric(imu[column], errors="coerce")
    imu = (
        imu.dropna(subset=required)
        .sort_values("t_ms")
        .drop_duplicates("t_ms")
        .reset_index(drop=True)
    )
    if len(imu) < 3:
        raise ValueError("At least three valid IMU samples are required")
    imu.insert(1, "elapsed_s", (imu["t_ms"] - imu["t_ms"].iloc[0]) / 1000.0)

    medium_path = data_dir / f"{log_id}_medium.csv"
    speed_source = None
    if medium_path.exists():
        medium = pd.read_csv(medium_path, usecols=lambda c: c in {"t_ms", "speed_mph"})
        if {"t_ms", "speed_mph"}.issubset(medium.columns):
            medium["t_ms"] = pd.to_numeric(medium["t_ms"], errors="coerce")
            medium["speed_mph"] = pd.to_numeric(medium["speed_mph"], errors="coerce")
            medium = medium.dropna().sort_values("t_ms").drop_duplicates("t_ms")
            if not medium.empty:
                imu["speed_mph"] = np.interp(
                    imu["t_ms"], medium["t_ms"], medium["speed_mph"],
                    left=medium["speed_mph"].iloc[0],
                    right=medium["speed_mph"].iloc[-1],
                )
                speed_source = medium_path
    return imu, imu_path, speed_source


def normalize(vector, name="vector"):
    vector = np.asarray(vector, dtype=float)
    magnitude = np.linalg.norm(vector)
    if not np.isfinite(magnitude) or magnitude <= 1e-12:
        raise ValueError(f"Cannot normalize {name}; magnitude is {magnitude}")
    return vector / magnitude


def rotation_between(source, target):
    """Return a proper rotation matrix mapping source onto target."""
    source = normalize(source, "rotation source")
    target = normalize(target, "rotation target")
    cross = np.cross(source, target)
    dot = float(np.clip(np.dot(source, target), -1.0, 1.0))
    sine = np.linalg.norm(cross)
    if sine < 1e-12:
        if dot > 0:
            return np.eye(3)
        basis = np.eye(3)[np.argmin(np.abs(source))]
        axis = normalize(np.cross(source, basis), "180-degree rotation axis")
        return -np.eye(3) + 2.0 * np.outer(axis, axis)

    x, y, z = cross
    skew = np.array([[0.0, -z, y], [z, 0.0, -x], [-y, x, 0.0]])
    return np.eye(3) + skew + (skew @ skew) * ((1.0 - dot) / sine**2)


def contiguous_segments(mask, time_s):
    mask = np.asarray(mask, dtype=bool)
    time_s = np.asarray(time_s, dtype=float)
    starts = np.flatnonzero(mask & np.r_[True, ~mask[:-1]])
    ends = np.flatnonzero(mask & np.r_[~mask[1:], True])
    return [
        {
            "start_idx": int(start),
            "end_idx": int(end),
            "start_s": float(time_s[start]),
            "end_s": float(time_s[end]),
            "duration_s": float(time_s[end] - time_s[start]),
            "samples": int(end - start + 1),
        }
        for start, end in zip(starts, ends)
    ]


def select_calibration_window(df):
    elapsed = df["elapsed_s"].to_numpy(dtype=float)
    if CALIBRATION_TIME_RANGE_S is not None:
        start_s, end_s = CALIBRATION_TIME_RANGE_S
        mask = (elapsed >= start_s) & (elapsed <= end_s)
        source = "manual elapsed-time range"
    elif "speed_mph" in df.columns:
        stationary = df["speed_mph"].abs().to_numpy() <= STATIONARY_SPEED_THRESHOLD_MPH
        candidates = [
            segment for segment in contiguous_segments(stationary, elapsed)
            if segment["duration_s"] >= MIN_CALIBRATION_DURATION_S
        ]
        if not candidates:
            raise ValueError("No sufficiently long zero-speed calibration window was found")
        selected = candidates[0]
        end_s = min(selected["start_s"] + CALIBRATION_MAX_DURATION_S, selected["end_s"])
        mask = (elapsed >= selected["start_s"]) & (elapsed <= end_s)
        source = f"first |speed| <= {STATIONARY_SPEED_THRESHOLD_MPH:g} mph segment"
    else:
        dt = np.median(np.diff(elapsed))
        window_samples = max(3, int(round(CALIBRATION_MAX_DURATION_S / dt)))
        accel_std = df[ACCEL_COLS].rolling(window_samples).std().pow(2).sum(axis=1)
        gyro_std = df[GYRO_COLS].rolling(window_samples).std().pow(2).sum(axis=1)
        score = accel_std + gyro_std / 10_000.0
        end_idx = int(score.idxmin())
        start_idx = end_idx - window_samples + 1
        mask = np.zeros(len(df), dtype=bool)
        mask[max(0, start_idx): end_idx + 1] = True
        source = "quietest fixed-duration IMU window (no speed stream)"

    mask = np.asarray(mask, dtype=bool)
    if mask.sum() < 3:
        raise ValueError("Calibration window contains fewer than three samples")
    duration_s = elapsed[mask][-1] - elapsed[mask][0]
    if duration_s < MIN_CALIBRATION_DURATION_S:
        raise ValueError(f"Calibration window is only {duration_s:.3f} s long")
    return mask, source


def estimate_forward(calibrated_linear_g, df, calibration_mask, gravity_rotation):
    horizontal = calibrated_linear_g.copy()
    horizontal[:, 2] = 0.0

    if FORWARD_IMU_VECTOR is not None:
        candidate = gravity_rotation @ np.asarray(FORWARD_IMU_VECTOR, dtype=float)
        candidate[2] = 0.0
        return normalize(candidate, "configured forward vector"), "configured raw-IMU vector", np.nan, None

    speed_accel = None
    if "speed_mph" in df.columns and df["speed_mph"].max() > STATIONARY_SPEED_THRESHOLD_MPH:
        elapsed = df["elapsed_s"].to_numpy(dtype=float)
        dt = np.median(np.diff(elapsed))
        smooth_samples = max(3, int(round(0.5 / dt)))
        if smooth_samples % 2 == 0:
            smooth_samples += 1
        speed_mps = df["speed_mph"].to_numpy(dtype=float) * MPH_TO_MPS
        speed_smooth = (
            pd.Series(speed_mps)
            .rolling(smooth_samples, center=True, min_periods=1)
            .mean()
            .to_numpy()
        )
        speed_accel = np.gradient(speed_smooth, elapsed)
        usable = (
            ~calibration_mask
            & np.isfinite(speed_accel)
            & (np.abs(speed_accel) >= MIN_SPEED_ACCEL_MPS2)
        )
        if usable.sum() >= 20:
            response = speed_accel[usable]
            coefficient = np.sum(horizontal[usable] * response[:, None], axis=0) / np.sum(response**2)
            if np.linalg.norm(coefficient) > 1e-9:
                forward = normalize(coefficient, "speed-regression forward vector")
                longitudinal_mps2 = (horizontal[usable] @ forward) * 9.80665
                correlation = float(np.corrcoef(longitudinal_mps2, response)[0, 1])
                return forward, "horizontal acceleration vs. d(speed)/dt regression", correlation, speed_accel

    active = (~calibration_mask) & np.isfinite(horizontal).all(axis=1)
    samples = horizontal[active]
    samples = samples[np.linalg.norm(samples, axis=1) > 1e-4]
    if len(samples) < 2:
        raise ValueError("Not enough dynamic acceleration to estimate vehicle forward")
    _, _, vh = np.linalg.svd(samples - samples.mean(axis=0), full_matrices=False)
    forward = normalize(vh[0], "PCA forward vector")
    raw_x_in_calibrated_frame = gravity_rotation @ np.array([1.0, 0.0, 0.0])
    raw_x_in_calibrated_frame[2] = 0.0
    if np.linalg.norm(raw_x_in_calibrated_frame) > 1e-9 and np.dot(forward, raw_x_in_calibrated_frame) < 0:
        forward = -forward
    return forward, "horizontal-acceleration PCA (sign toward raw IMU +x)", np.nan, speed_accel


def decorate_gg_axis(ax, limit_g):
    ax.axhline(0.0, color="0.35", linewidth=0.8)
    ax.axvline(0.0, color="0.35", linewidth=0.8)
    for radius in np.arange(0.5, limit_g + 0.01, 0.5):
        ax.add_patch(Circle((0, 0), radius, fill=False, color="0.55", linewidth=0.7, alpha=0.5))
    ax.set(xlim=(-limit_g, limit_g), ylim=(-limit_g, limit_g))
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlabel("Lateral acceleration (g, +left)")
    ax.set_ylabel("Longitudinal acceleration (g, +forward)")
```

## Load the IMU and select a stationary interval


```python
imu, imu_path, speed_path = load_imu_and_speed(DATA_DIR, LOG_ID)
calibration_mask, calibration_source = select_calibration_window(imu)
calibration = imu.loc[calibration_mask].copy()

print(f"Loaded {len(imu):,} IMU samples from {imu_path}")
print(f"Duration: {imu['elapsed_s'].iloc[-1]:.2f} s")
print(f"Speed stream: {speed_path if speed_path is not None else 'not available'}")
print(f"Calibration selection: {calibration_source}")
print(
    f"Calibration window: {calibration['elapsed_s'].iloc[0]:.3f} to "
    f"{calibration['elapsed_s'].iloc[-1]:.3f} s ({len(calibration):,} samples)"
)
```

## Calibrate gravity, gyro bias, and vehicle axes

The negative of the stationary accelerometer mean estimates physical gravity and is mapped to `[0, 0, -1] g`. The stationary accelerometer specific force therefore maps to `[0, 0, +1] g`; adding gravity yields approximately zero linear acceleration at rest. Vehicle lateral is positive left and longitudinal is positive forward.


```python
raw_accel_g = imu[ACCEL_COLS].to_numpy(dtype=float)
raw_gyro_dps = imu[GYRO_COLS].to_numpy(dtype=float)
stationary_accel_mean_g = calibration[ACCEL_COLS].mean().to_numpy(dtype=float)
stationary_gravity_raw_g = -stationary_accel_mean_g
stationary_gyro_bias_dps = calibration[GYRO_COLS].mean().to_numpy(dtype=float)
measured_gravity_g = np.linalg.norm(stationary_accel_mean_g)
accel_scale = 1.0 / measured_gravity_g
target_down = np.array([0.0, 0.0, -1.0])
gravity_rotation = rotation_between(stationary_gravity_raw_g, target_down)

calibrated_accel_g = (gravity_rotation @ (raw_accel_g * accel_scale).T).T
linear_accel_g = calibrated_accel_g + target_down
calibrated_gyro_dps = raw_gyro_dps - stationary_gyro_bias_dps

forward_cal, forward_method, forward_correlation, speed_accel_mps2 = estimate_forward(
    linear_accel_g, imu, calibration_mask, gravity_rotation
)
forward_cal[2] = 0.0
forward_cal = normalize(forward_cal, "horizontal forward")
up_cal = -target_down
left_cal = normalize(np.cross(up_cal, forward_cal), "vehicle left")
down_cal = target_down.copy()

imu_calibrated = imu.copy()
for column, values in zip(["accel_forward_g", "accel_left_g", "accel_up_g"], [
    linear_accel_g @ forward_cal,
    linear_accel_g @ left_cal,
    linear_accel_g @ up_cal,
]):
    imu_calibrated[column] = values
for axis_index, axis in enumerate(AXES):
    imu_calibrated[f"accel_cal_{axis}_g"] = calibrated_accel_g[:, axis_index]
    imu_calibrated[f"gyro_cal_{axis}_dps"] = calibrated_gyro_dps[:, axis_index]

finite_gg = np.isfinite(imu_calibrated[["accel_left_g", "accel_forward_g"]]).all(axis=1)
gg_data = imu_calibrated.loc[finite_gg].copy()
robust_extent = np.nanpercentile(
    np.abs(gg_data[["accel_left_g", "accel_forward_g"]].to_numpy()), 99.5
)
gg_limit_g = float(GG_LIMIT_G) if GG_LIMIT_G is not None else max(0.5, np.ceil(robust_extent * 10) / 10)
```

## Calibration metrics and forward direction


```python
stationary_calibrated = calibrated_accel_g[calibration_mask]
stationary_linear = linear_accel_g[calibration_mask]
stationary_gyro_cal = calibrated_gyro_dps[calibration_mask]
rotation_angle_deg = np.degrees(
    np.arccos(np.clip((np.trace(gravity_rotation) - 1.0) / 2.0, -1.0, 1.0))
)

summary_metrics = pd.DataFrame(
    [
        ("Calibration samples", int(calibration_mask.sum()), "samples"),
        ("Calibration duration", calibration['elapsed_s'].iloc[-1] - calibration['elapsed_s'].iloc[0], "s"),
        ("Measured stationary magnitude", measured_gravity_g, "g"),
        ("Common accelerometer scale", accel_scale, "multiplier"),
        ("Gravity alignment rotation", rotation_angle_deg, "deg"),
        ("Post-calibration gravity magnitude", np.linalg.norm(-stationary_calibrated.mean(axis=0)), "g"),
        ("Stationary linear-acceleration RMS", np.sqrt(np.mean(stationary_linear**2)), "g"),
        ("Stationary calibrated-gyro RMS", np.sqrt(np.mean(stationary_gyro_cal**2)), "deg/s"),
        ("Forward/speed-acceleration correlation", forward_correlation, "Pearson r"),
    ],
    columns=["metric", "value", "units"],
).set_index("metric")

axis_metrics = pd.DataFrame(
    {
        "raw stationary mean (g)": stationary_accel_mean_g,
        "raw stationary std (g)": calibration[ACCEL_COLS].std(ddof=1).to_numpy(),
        "calibrated specific-force mean (g)": stationary_calibrated.mean(axis=0),
        "calibrated stationary std (g)": stationary_calibrated.std(axis=0, ddof=1),
        "linear stationary mean (g)": stationary_linear.mean(axis=0),
        "gyro bias removed (deg/s)": stationary_gyro_bias_dps,
    },
    index=["x", "y", "z"],
)
rotation_table = pd.DataFrame(
    gravity_rotation, index=["cal x", "cal y", "cal z"], columns=["raw x", "raw y", "raw z"]
)

forward_raw = normalize(gravity_rotation.T @ forward_cal, "raw-frame forward")
left_raw = normalize(gravity_rotation.T @ left_cal, "raw-frame left")
up_raw = normalize(gravity_rotation.T @ up_cal, "raw-frame up")
down_raw = -up_raw
direction_table = pd.DataFrame(
    {
        "forward": forward_raw,
        "left": left_raw,
        "up": up_raw,
        "down / gravity": down_raw,
    },
    index=["raw IMU x", "raw IMU y", "raw IMU z"],
)

print(f"Forward-direction method: {forward_method}")
if np.isfinite(forward_correlation):
    print(f"Forward vs. d(speed)/dt correlation: {forward_correlation:.3f}")
print("Forward direction in raw IMU coordinates: " + np.array2string(forward_raw, precision=6))
print("\nCalibration summary")
display(summary_metrics)
print("Per-axis stationary metrics")
display(axis_metrics)
print("Gravity-alignment rotation (calibrated = R @ scaled_raw)")
display(rotation_table)
print("Vehicle basis expressed in raw IMU coordinates")
display(direction_table)
```

## Magma G-G density heatmap

Zero-count bins are transparent. The color scale clips at `HEATMAP_SATURATION_COUNT` (default: 8 samples), deliberately making low-density structure visible.


```python
histogram, lateral_edges, longitudinal_edges = np.histogram2d(
    gg_data["accel_left_g"],
    gg_data["accel_forward_g"],
    bins=HEATMAP_BINS,
    range=[[-gg_limit_g, gg_limit_g], [-gg_limit_g, gg_limit_g]],
)
masked_histogram = np.ma.masked_equal(histogram.T, 0)
magma = plt.colormaps["magma"].copy()
magma.set_bad(alpha=0.0)

fig, ax = plt.subplots(figsize=(9, 8))
mesh = ax.pcolormesh(
    lateral_edges,
    longitudinal_edges,
    masked_histogram,
    cmap=magma,
    norm=Normalize(vmin=1, vmax=HEATMAP_SATURATION_COUNT, clip=True),
    shading="flat",
)
decorate_gg_axis(ax, gg_limit_g)
ax.set_title(
    f"{LOG_ID} calibrated IMU G-G density | magma saturates at "
    f"{HEATMAP_SATURATION_COUNT} samples/bin"
)
colorbar = fig.colorbar(mesh, ax=ax, label="Samples per bin", shrink=0.82, extend="max")
colorbar.set_ticks(np.arange(1, HEATMAP_SATURATION_COUNT + 1))
plt.tight_layout()
plt.show()
```

## Gap-aware sustained-g analysis

The original direction, scatter, and magma heatmap above are intentionally retained. This additional analysis estimates pitch/roll gravity motion from the bias-corrected gyro, anchors that gravity estimate whenever the car is stationary, rejects accelerometer rail hits, splits the signal at timestamp gaps, applies a zero-phase 5 Hz low-pass filter, and reports both filtered peaks and 250 ms sustained peaks.

The single-pose calibration used by this notebook corrects a common scale factor. For final vehicle numbers, replace it with a six-position accelerometer calibration and set `FORWARD_IMU_VECTOR` from the measured mounting orientation.


```python
def gyro_aided_specific_gravity(
    accel_g, gyro_dps, time_s, stationary_mask, initial_mask, max_gap_s, correction_tau_s
):
    """Track the +1 g stationary accelerometer vector in vehicle coordinates.

    Gyro propagation handles pitch/roll while accelerometer corrections are only
    accepted when the vehicle is stationary, so sustained cornering is not
    incorrectly interpreted as attitude.
    """
    accel_g = np.asarray(accel_g, dtype=float)
    gyro_radps = np.deg2rad(np.asarray(gyro_dps, dtype=float))
    time_s = np.asarray(time_s, dtype=float)
    stationary_mask = np.asarray(stationary_mask, dtype=bool)
    initial_mask = np.asarray(initial_mask, dtype=bool)
    gravity_g = np.full_like(accel_g, np.nan)

    initial = normalize(np.nanmean(accel_g[initial_mask], axis=0), "initial gravity vector")
    gravity_g[0] = initial
    nominal_dt_s = float(np.median(np.diff(time_s)[np.diff(time_s) > 0]))

    for index in range(1, len(time_s)):
        dt_s = time_s[index] - time_s[index - 1]
        previous = gravity_g[index - 1]
        if np.isfinite(dt_s) and 0 < dt_s <= max_gap_s:
            omega = 0.5 * (gyro_radps[index - 1] + gyro_radps[index])
            predicted = previous - np.cross(omega, previous) * dt_s
            predicted = normalize(predicted, "gyro-propagated gravity")
        else:
            predicted = previous

        accel_norm = np.linalg.norm(accel_g[index])
        if stationary_mask[index] and np.isfinite(accel_norm) and 0.5 < accel_norm < 1.5:
            observed = accel_g[index] / accel_norm
            blend_dt_s = dt_s if np.isfinite(dt_s) and dt_s > 0 else nominal_dt_s
            blend = 1.0 - np.exp(-blend_dt_s / correction_tau_s)
            predicted = normalize((1.0 - blend) * predicted + blend * observed, "corrected gravity")
        gravity_g[index] = predicted

    return gravity_g


def contiguous_valid_slices(time_s, valid_mask, max_gap_s):
    time_s = np.asarray(time_s, dtype=float)
    valid_indices = np.flatnonzero(np.asarray(valid_mask, dtype=bool))
    if not len(valid_indices):
        return []
    breaks = (np.diff(valid_indices) > 1) | (np.diff(time_s[valid_indices]) > max_gap_s)
    groups = np.split(valid_indices, np.flatnonzero(breaks) + 1)
    return [slice(int(group[0]), int(group[-1]) + 1) for group in groups if len(group)]


def filter_and_average_by_segment(time_s, values_g, valid_mask):
    time_s = np.asarray(time_s, dtype=float)
    values_g = np.asarray(values_g, dtype=float)
    positive_dt_s = np.diff(time_s)
    positive_dt_s = positive_dt_s[positive_dt_s > 0]
    nominal_dt_s = float(np.median(positive_dt_s))
    sample_rate_hz = 1.0 / nominal_dt_s
    gap_threshold_s = MAX_SAMPLE_GAP_FACTOR * nominal_dt_s
    window_samples = max(1, int(round(SUSTAINED_WINDOW_S * sample_rate_hz)))
    sos = butter(4, FILTER_CUTOFF_HZ, btype="lowpass", fs=sample_rate_hz, output="sos")

    filtered_g = np.full_like(values_g, np.nan)
    sustained_g = np.full_like(values_g, np.nan)
    accepted_segments = []
    minimum_samples = max(16, window_samples, int(round(MIN_FILTER_SEGMENT_S * sample_rate_hz)))
    for segment in contiguous_valid_slices(time_s, valid_mask, gap_threshold_s):
        if segment.stop - segment.start < minimum_samples:
            continue
        segment_filtered = sosfiltfilt(sos, values_g[segment], axis=0)
        filtered_g[segment] = segment_filtered
        sustained_g[segment] = (
            pd.DataFrame(segment_filtered)
            .rolling(window_samples, center=True, min_periods=window_samples)
            .mean()
            .to_numpy()
        )
        accepted_segments.append(segment)
    return filtered_g, sustained_g, accepted_segments, sample_rate_hz, gap_threshold_s, window_samples


elapsed_s = imu_calibrated["elapsed_s"].to_numpy(dtype=float)
gyro_vehicle_dps = (gravity_rotation @ calibrated_gyro_dps.T).T
if "speed_mph" in imu_calibrated.columns:
    stationary_attitude_mask = (
        imu_calibrated["speed_mph"].abs().to_numpy(dtype=float)
        <= STATIONARY_SPEED_THRESHOLD_MPH
    )
    moving_mask = (
        imu_calibrated["speed_mph"].abs().to_numpy(dtype=float)
        >= MOVING_SPEED_THRESHOLD_MPH
    )
    motion_source = f"|speed| >= {MOVING_SPEED_THRESHOLD_MPH:g} mph"
else:
    stationary_attitude_mask = calibration_mask.copy()
    moving_mask = ~calibration_mask
    motion_source = "all samples outside the calibration window"

positive_dt_s = np.diff(elapsed_s)
positive_dt_s = positive_dt_s[positive_dt_s > 0]
preliminary_gap_s = MAX_SAMPLE_GAP_FACTOR * float(np.median(positive_dt_s))
gyro_specific_gravity_g = gyro_aided_specific_gravity(
    calibrated_accel_g,
    gyro_vehicle_dps,
    elapsed_s,
    stationary_attitude_mask,
    calibration_mask,
    preliminary_gap_s,
    ATTITUDE_STATIONARY_CORRECTION_TAU_S,
)
gyro_gravity_tilt_deg = np.degrees(
    np.arccos(np.clip(gyro_specific_gravity_g @ up_cal, -1.0, 1.0))
)
gyro_gravity_tilt_p99_deg = float(np.nanpercentile(gyro_gravity_tilt_deg, 99))
if gyro_gravity_tilt_p99_deg <= MAX_REASONABLE_GRAVITY_TILT_DEG:
    specific_gravity_g = gyro_specific_gravity_g
    gravity_method = "gyro propagation with stationary accelerometer updates"
    gravity_fallback_used = False
else:
    fixed_gravity_g = normalize(
        calibrated_accel_g[calibration_mask].mean(axis=0), "fixed stationary gravity"
    )
    specific_gravity_g = np.tile(fixed_gravity_g, (len(calibrated_accel_g), 1))
    gravity_method = "fixed stationary vector (gyro-attitude plausibility fallback)"
    gravity_fallback_used = True
attitude_compensated_linear_g = calibrated_accel_g - specific_gravity_g
vehicle_planar_g = np.column_stack([
    attitude_compensated_linear_g @ forward_cal,
    attitude_compensated_linear_g @ left_cal,
])

finite_sensor_mask = np.isfinite(raw_accel_g).all(axis=1) & np.isfinite(vehicle_planar_g).all(axis=1)
saturation_mask = np.any(np.abs(raw_accel_g) >= ACCEL_SATURATION_G, axis=1)
valid_filter_mask = finite_sensor_mask & ~saturation_mask
(
    filtered_planar_g, sustained_planar_g, filter_segments, nominal_sample_rate_hz,
    gap_threshold_s, sustained_window_samples,
) = filter_and_average_by_segment(elapsed_s, vehicle_planar_g, valid_filter_mask)

imu_performance = imu_calibrated.copy()
imu_performance["gravity_forward_g"] = specific_gravity_g @ forward_cal
imu_performance["gravity_left_g"] = specific_gravity_g @ left_cal
imu_performance["accel_forward_attitude_g"] = vehicle_planar_g[:, 0]
imu_performance["accel_left_attitude_g"] = vehicle_planar_g[:, 1]
imu_performance["accel_forward_filtered_g"] = filtered_planar_g[:, 0]
imu_performance["accel_left_filtered_g"] = filtered_planar_g[:, 1]
imu_performance["accel_forward_sustained_g"] = sustained_planar_g[:, 0]
imu_performance["accel_left_sustained_g"] = sustained_planar_g[:, 1]
imu_performance["sensor_saturated"] = saturation_mask
imu_performance["moving_sample"] = moving_mask

performance_mask = moving_mask & np.isfinite(sustained_planar_g).all(axis=1)
if not performance_mask.any():
    raise ValueError("No valid moving samples remain after sustained-g processing")

gap_count = int(np.sum(np.diff(elapsed_s) > gap_threshold_s))
quality_summary = pd.DataFrame(
    [
        ("Nominal sample rate", nominal_sample_rate_hz, "Hz"),
        ("Gap threshold", gap_threshold_s * 1000.0, "ms"),
        ("Timestamp gaps split", gap_count, "gaps"),
        ("Accepted filter segments", len(filter_segments), "segments"),
        ("Accelerometer rail hits rejected", int(saturation_mask.sum()), "samples"),
        ("Stationary attitude updates", int(stationary_attitude_mask.sum()), "samples"),
        ("Gyro-inferred gravity tilt, 99th percentile", gyro_gravity_tilt_p99_deg, "deg"),
        ("Gravity plausibility limit", MAX_REASONABLE_GRAVITY_TILT_DEG, "deg"),
        ("Sustained averaging window", sustained_window_samples, "samples"),
        ("Valid moving sustained samples", int(performance_mask.sum()), "samples"),
    ],
    columns=["quality check", "value", "units"],
).set_index("quality check")

print(f"Performance sample selection: {motion_source}")
print(f"Gravity compensation: {gravity_method}")
if gravity_fallback_used:
    print(
        f"Warning: gyro-aided gravity tilt reached {gyro_gravity_tilt_p99_deg:.1f} deg at the "
        f"99th percentile, above the {MAX_REASONABLE_GRAVITY_TILT_DEG:g} deg limit; "
        "using the stationary gravity vector for performance metrics."
    )
if not 0.98 <= measured_gravity_g <= 1.02:
    print(
        f"Warning: stationary magnitude was {measured_gravity_g:.4f} g before the common-scale "
        "correction; perform a six-position calibration before treating these as final results."
    )
display(quality_summary)
```

```python
def directional_peak(values, direction, mask):
    score = direction * np.asarray(values, dtype=float)
    score = np.where(mask, score, np.nan)
    index = int(np.nanargmax(score))
    return float(score[index]), index


filtered_forward = imu_performance["accel_forward_filtered_g"].to_numpy(dtype=float)
filtered_left = imu_performance["accel_left_filtered_g"].to_numpy(dtype=float)
sustained_forward = imu_performance["accel_forward_sustained_g"].to_numpy(dtype=float)
sustained_left = imu_performance["accel_left_sustained_g"].to_numpy(dtype=float)
metric_specs = [
    ("Maximum acceleration", filtered_forward, sustained_forward, +1.0),
    ("Maximum braking", filtered_forward, sustained_forward, -1.0),
    ("Maximum left cornering", filtered_left, sustained_left, +1.0),
    ("Maximum right cornering", filtered_left, sustained_left, -1.0),
]
metric_rows = []
peak_indices = {}
for label, filtered_values, sustained_values, direction in metric_specs:
    filtered_peak_g, _ = directional_peak(filtered_values, direction, performance_mask)
    sustained_peak_g, peak_index = directional_peak(sustained_values, direction, performance_mask)
    peak_indices[label] = peak_index
    metric_rows.append(
        {
            "metric": label,
            "5 Hz peak (g)": filtered_peak_g,
            "250 ms sustained (g)": sustained_peak_g,
            "time (s)": elapsed_s[peak_index],
            "forward at sustained peak (g)": sustained_forward[peak_index],
            "left at sustained peak (g)": sustained_left[peak_index],
            "speed at sustained peak (mph)": (
                imu_performance["speed_mph"].iloc[peak_index]
                if "speed_mph" in imu_performance.columns else np.nan
            ),
        }
    )

filtered_combined = np.hypot(filtered_forward, filtered_left)
sustained_combined = np.hypot(sustained_forward, sustained_left)
filtered_combined_peak_g, _ = directional_peak(filtered_combined, +1.0, performance_mask)
sustained_combined_peak_g, combined_peak_index = directional_peak(
    sustained_combined, +1.0, performance_mask
)
peak_indices["Maximum combined"] = combined_peak_index
metric_rows.append(
    {
        "metric": "Maximum combined",
        "5 Hz peak (g)": filtered_combined_peak_g,
        "250 ms sustained (g)": sustained_combined_peak_g,
        "time (s)": elapsed_s[combined_peak_index],
        "forward at sustained peak (g)": sustained_forward[combined_peak_index],
        "left at sustained peak (g)": sustained_left[combined_peak_index],
        "speed at sustained peak (mph)": (
            imu_performance["speed_mph"].iloc[combined_peak_index]
            if "speed_mph" in imu_performance.columns else np.nan
        ),
    }
)
performance_summary = pd.DataFrame(metric_rows).set_index("metric")
print("Peak values are positive magnitudes; the forward/left columns retain axis signs.")
display(performance_summary)
```

## Improved sustained G-G heatmap

This additional heatmap uses only moving, valid 250 ms sustained samples. A logarithmic count scale shows both the frequently occupied center and the sparse performance envelope without clipping every bin above an arbitrary low count. The overlaid line is the 99th-percentile radius in angular sectors, and the markers locate the reported sustained extrema.


```python
def radial_quantile_envelope(lateral_g, forward_g, angular_bins=72, quantile=0.99, minimum_samples=10):
    lateral_g = np.asarray(lateral_g, dtype=float)
    forward_g = np.asarray(forward_g, dtype=float)
    angle = np.arctan2(forward_g, lateral_g)
    radius = np.hypot(lateral_g, forward_g)
    edges = np.linspace(-np.pi, np.pi, angular_bins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
    envelope_radius = np.full(angular_bins, np.nan)
    for bin_index in range(angular_bins):
        in_bin = (angle >= edges[bin_index]) & (angle < edges[bin_index + 1])
        if in_bin.sum() >= minimum_samples:
            envelope_radius[bin_index] = np.quantile(radius[in_bin], quantile)
    return envelope_radius * np.cos(centers), envelope_radius * np.sin(centers)


sustained_gg = imu_performance.loc[
    performance_mask, ["accel_left_sustained_g", "accel_forward_sustained_g"]
].copy()
sustained_lateral = sustained_gg["accel_left_sustained_g"].to_numpy(dtype=float)
sustained_longitudinal = sustained_gg["accel_forward_sustained_g"].to_numpy(dtype=float)
sustained_extent = np.max(np.abs(np.column_stack([sustained_lateral, sustained_longitudinal])))
improved_limit_g = (
    float(IMPROVED_GG_LIMIT_G)
    if IMPROVED_GG_LIMIT_G is not None
    else max(0.5, np.ceil(sustained_extent * 10.0) / 10.0)
)

improved_histogram, improved_lateral_edges, improved_longitudinal_edges = np.histogram2d(
    sustained_lateral,
    sustained_longitudinal,
    bins=IMPROVED_HEATMAP_BINS,
    range=[[-improved_limit_g, improved_limit_g], [-improved_limit_g, improved_limit_g]],
)
masked_improved_histogram = np.ma.masked_equal(improved_histogram.T, 0)
maximum_bin_count = float(improved_histogram.max())
density_norm = LogNorm(vmin=1.0, vmax=100)
envelope_left_g, envelope_forward_g = radial_quantile_envelope(
    sustained_lateral, sustained_longitudinal
)

fig, ax = plt.subplots(figsize=(10, 8.5))
mesh = ax.pcolormesh(
    improved_lateral_edges,
    improved_longitudinal_edges,
    masked_improved_histogram,
    cmap="magma",
    norm=density_norm,
    shading="flat",
)
ax.plot(
    envelope_left_g, envelope_forward_g, color="cyan", linewidth=1.8,
    label="99th-percentile radial envelope",
)
peak_styles = {
    "Maximum acceleration": ("^", "lime"),
    "Maximum braking": ("v", "deepskyblue"),
    "Maximum left cornering": ("<", "white"),
    "Maximum right cornering": (">", "gold"),
    "Maximum combined": ("*", "red"),
}
for label, peak_index in peak_indices.items():
    marker, color = peak_styles[label]
    ax.scatter(
        sustained_left[peak_index], sustained_forward[peak_index],
        marker=marker, s=95 if marker != "*" else 150, color=color,
        edgecolor="black", linewidth=0.7, zorder=5, label=label,
    )
decorate_gg_axis(ax, improved_limit_g)
ax.set_title(
    f"{LOG_ID} sustained G-G density | {FILTER_CUTOFF_HZ:g} Hz low-pass, "
    f"{SUSTAINED_WINDOW_S * 1000:.0f} ms average"
)
fig.colorbar(mesh, ax=ax, label="Moving sustained samples per bin (log scale)", shrink=0.84)
ax.legend(loc="upper left", fontsize=8, framealpha=0.9)
plt.tight_layout()
plt.show()
```

### Reading the plots

- Right/left position is lateral acceleration; positive values are left turns.
- Up/down position is longitudinal acceleration; positive values are acceleration in the estimated forward direction.
- The scatter color shows when a sample occurred.
- The original heatmap shows occupancy, with bins at or above the configured 5–10 sample saturation count sharing the brightest color.
- The improved heatmap is the preferred performance view: it excludes stationary samples and sensor rail hits, handles timestamp gaps, and uses 250 ms sustained acceleration with logarithmic density.
- Treat the peak table as provisional until a six-position accelerometer calibration and measured mounting transform are available.
- If the reported forward method is the PCA fallback, verify the direction against the physical sensor mounting and set `FORWARD_IMU_VECTOR` when necessary.
